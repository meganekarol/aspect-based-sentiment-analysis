```python
import random
import json
import pandas as pd
```

```python
sample_reviews = []
with open("../data/yelp_sample_20k.json", "r", encoding="utf-8") as f:
    all_lines = f.readlines()
    sample_lines = random.sample(all_lines, 60)  # pick 5 random reviews
    for line in sample_lines:
     data = json.loads(line)
     sample_reviews.append(data)
```

```python
df = pd.DataFrame(sample_reviews)[["text", "stars"]]
df.to_json("../data/yelp_sample.json", orient="records", lines=True)
```
